```python
import random
import functools


LAST_NAME = "LINSANGAN"    
SEED_NUM = 6           
FAVORITE_ARTIST = "STEVE PERRY" 


# ==============================================================================
# PROBLEM 2: RECURSIVE FAULT TRACE
# ==============================================================================

fault_execution_log = []

def trace_logger(func):
    """Decorator to record function calls in the execution log."""
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        log_msg = f"[LOG] Executing '{func.__name__}' with code={args[0]} at depth={args[1] if len(args) > 1 else 1}"
        fault_execution_log.append(log_msg)
        return func(*args, **kwargs)
    return wrapper

def generate_fault_code(surname: str, seed_digit: int, artist: str) -> int:
    """Requirement 1: Generate student-specific fault code."""
    return (sum(ord(c) for c in surname) * seed_digit) + len(artist)

@trace_logger
def trace_fault_recursive(fault_code: int, level: int = 1, trace_sequence: list = None) -> tuple:
    """Requirements 2, 3, 4: Recursive function with base condition and counter."""
    if trace_sequence is None:
        trace_sequence = []

    trace_sequence.append(f"Level {level}: Fault Code {fault_code}")

    # Base condition (Requirement 3)
    if fault_code <= 1:
        # Total recursive calls made = level - 1 (Requirement 4)
        return trace_sequence, level - 1

    next_code = fault_code // 2 if fault_code % 2 == 0 else fault_code - 1
    return trace_fault_recursive(next_code, level + 1, trace_sequence)


# ==============================================================================
# PROBLEM 3: INTELLIGENT EQUIPMENT MONITORING PIPELINE
# ==============================================================================

pipeline_log = []

def pipeline_monitor(func):
    """Requirement 5: Decorator to monitor major processing function."""
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        pipeline_log.append(f"[LOG] Pipeline monitoring started on '{func.__name__}'.")
        result = func(*args, **kwargs)
        pipeline_log.append(f"[LOG] Pipeline monitoring completed on '{func.__name__}'.")
        return result
    return wrapper

# Requirement 1: Generate student-specific telemetry data stream
def generate_telemetry_stream(surname: str, seed_digit: int, artist: str, count: int = 8):
    """Requirement 3: Generator function yielding telemetry readings[cite: 2]."""
    seed_val = sum(ord(c) for c in surname) + seed_digit + len(artist)
    random.seed(seed_val)
    
    # Stream includes valid values, out-of-bounds numbers, and corrupted strings
    sample_values = [72.5, 125.0, "ERR_SENSOR", 45.0, 95.0, -10.0, "MALFUNCTION", 88.0]
    for i in range(min(count, len(sample_values))):
        yield sample_values[i]

@pipeline_monitor
def process_telemetry_stream(stream):
    """Requirements 4, 6, 7: Main pipeline processing telemetry continuous stream[cite: 2]."""
    processed_count = 0
    valid_readings = []
    invalid_readings = []
    abnormal_conditions = []

    # Requirement 4: Lambda function to calibrate or transform telemetry data[cite: 2]
    calibrate = lambda val: round(val * 1.02, 2)

    for raw_val in stream:
        processed_count += 1
        # Requirement 6: Exception handling for invalid/unexpected telemetry[cite: 2]
        try:
            if not isinstance(raw_val, (int, float)):
                raise TypeError(f"Invalid telemetry format: '{raw_val}'")
            if raw_val < 0 or raw_val > 100:
                raise ValueError(f"Out of bounds reading: {raw_val}")

            calibrated_val = calibrate(raw_val)
            valid_readings.append(calibrated_val)

            # Detect abnormal condition (> 90)
            if calibrated_val > 90.0:
                abnormal_conditions.append(calibrated_val)

        except (TypeError, ValueError) as e:
            invalid_readings.append(f"Reading #{processed_count} ({raw_val}): {e}")

    # Requirement 7: Recursive analysis of abnormal conditions[cite: 2]
    def analyze_abnormal_recursive(vals, depth=1):
        if not vals or depth > len(vals):
            return f"Recursive trace reached base condition at depth {depth}."
        pipeline_log.append(f"[LOG] Recursive trace analyzing abnormal value {vals[depth-1]} at depth {depth}")
        return analyze_abnormal_recursive(vals, depth + 1)

    recursive_trace_msg = analyze_abnormal_recursive(abnormal_conditions)

    # Determine overall equipment status
    if len(abnormal_conditions) > 1 or len(invalid_readings) > 2:
        overall_status = "CRITICAL - IMMEDIATE INSPECTION REQUIRED"
    elif len(abnormal_conditions) == 1:
        overall_status = "WARNING - ELEVATED READINGS DETECTED"
    else:
        overall_status = "NORMAL OPERATING CONDITION"

    # Requirement 8: Return structured final report dictionary[cite: 2]
    return {
        "processed_count": processed_count,
        "valid_count": len(valid_readings),
        "invalid_count": len(invalid_readings),
        "invalid_details": invalid_readings,
        "abnormal_conditions": abnormal_conditions,
        "recursive_trace": recursive_trace_msg,
        "overall_status": overall_status
    }


# ==============================================================================
# MAIN EXECUTION & ASSESSMENT DATA OUTPUTS
# ==============================================================================

if __name__ == "__main__":
    # --- PROBLEM 2 RUN ---
    fault_code = generate_fault_code(LAST_NAME, SEED_NUM, FAVORITE_ARTIST)
    trace_seq, call_count = trace_fault_recursive(fault_code)

    print("=" * 65)
    print("2. RECURSIVE FAULT TRACE RESULTS")
    print("=" * 65)
    print(f"Generated Fault Data: {fault_code}\n")
    print("Recursive Trace:")
    for step in trace_seq:
        print(f"  -> {step}")
    print(f"\nNumber of Recursive Calls: {call_count}\n")
    print("Execution Log:")
    for log in fault_execution_log[:5]:  # Display snippet of log
        print(f"  {log}")
    print("  ... [truncated for brevity]")
    print(f"\nFinal Output: Fault trace completed. Reached base condition in {call_count} recursive calls.")
    print("=" * 65)

    # --- PROBLEM 3 RUN ---
    telemetry_stream = generate_telemetry_stream(LAST_NAME, SEED_NUM, FAVORITE_ARTIST)
    report = process_telemetry_stream(telemetry_stream)

    print("\n" + "=" * 65)
    print("3. INTELLIGENT EQUIPMENT MONITORING PIPELINE RESULTS")
    print("=" * 65)
    print(f"Generated Telemetry Stream: Derived for [{LAST_NAME}_{SEED_NUM}_{FAVORITE_ARTIST}]\n")
    print("Pipeline Execution Log:")
    for log in pipeline_log:
        print(f"  {log}")
    print()
    print("Final Diagnostic Report:")
    print(f"  - Total Processed Readings : {report['processed_count']}")
    print(f"  - Valid Readings Count     : {report['valid_count']}")
    print(f"  - Invalid Readings Count   : {report['invalid_count']}")
    print(f"  - Detected Abnormalities   : {report['abnormal_conditions']}")
    print(f"  - Recursive Trace Status   : {report['recursive_trace']}")
    print(f"  - Overall Equipment Status : {report['overall_status']}")
    print("=" * 65)
```
